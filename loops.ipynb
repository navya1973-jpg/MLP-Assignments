```python
for i in range(1,1000,1):            #break statement
    if(i==8):
        break
    else:
        print(i)
```

```python
for i in range(1,15,1):            #continue statement
    if(i==8 or i==9 or i==10):
        continue
    else:
        print(i)
```

```python
for i in range(1,15,1):            #pass statement
    if(i==8 or i==9 or i==10):
        pass
    else:
        print(i)
```

```python
#print prime numbers between 2 to 20 using nested while
i=2
while(i<=20):
    j=2
    while(j<i):
        if(i%j==0):
            break
        j+=1
    else:
        print(i,end="\t")
    i+=1
```
