```python
#write a program to print the given number is prime or not
num=int(input("enter the number"))
count=0
for i in range(2,num,1):
    if (num%i==0):
        count=count+1
print("count=",count)
if(count==0):
        print(num,"is prime")
else:
    print(num,"is not prime")
```

```python
for i in range(1,1000,1):            #break statement
    if(i==8):
        break
    else:
        print(i)
```

```python
for i in range(1,15,1):            #continue statement
    if(i==8 or i==9 or i==10):
        continue
    else:
        print(i)
```

```python
for i in range(1,15,1):            #pass statement
    if(i==8 or i==9 or i==10):
        pass
    else:
        print(i)
```
