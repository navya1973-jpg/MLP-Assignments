```python
'''1. E-commerce Discount Calculator

Scenario: An online store offers discounts based on the purchase amount:

10% discount for purchases between $100 and $500

20% discount for purchases above $500

No discount for purchases below $100


Task: Write a program that takes the purchase amount as input and calculates the discount and final amount to be paid.'''

amount=float(input("enter purchase amount:$"))
if (amount<100):
    print("no discount")
elif (amount<=500):
    discount=amount*0.10
else:
    discount=amount*0.20

final_amount = amount - discount

print("purchase amount:$",amount)
print("discount:$",discount)
print("final amount:$",final_amount)
```

```python
'''14. Password Strength Checker
Accept a password.
If its length is at least 8 characters, check whether it contains a digit.
If it does, check whether it contains an uppercase letter. 
Classify it as strong or moderate based on these conditions. 
Otherwise, classify it as weak.'''
password = input("Enter password: ")

if len(password) >= 8:
    if any(char.isdigit() for char in password):
        if any(char.isupper() for char in password):
            print("Strong password")
        else:
            print("Moderate password")
    else:
        print("Moderate password")
else:
    print("Weak password")
```

```python
'''20. Mini Banking System
Accept account type, account balance, and transaction type (deposit or withdrawal). 
If the account type is savings, check the transaction type. For withdrawals, check minimum balance requirements; for deposits, validate the deposit amount.
If the account type is current, apply a different minimum balance rule.'''
account = input("Enter account type (savings/current): ").lower()
balance = float(input("Enter account balance: "))
transaction = input("Enter transaction (deposit/withdrawal): ").lower()
amount = float(input("Enter transaction amount: "))

if account == "savings":
    if transaction == "deposit":
        if amount > 0:
            balance = balance + amount
            print("Deposit successful")
        else:
            print("Invalid deposit amount")
    elif transaction == "withdrawal":
        if balance - amount >= 1000 and amount > 0:
            balance = balance - amount
            print("Withdrawal successful")
        else:
            print("Minimum balance requirement not met or invalid amount")
    else:
        print("Invalid transaction type")

elif account == "current":
    if transaction == "deposit":
        if amount > 0:
            balance = balance + amount
            print("Deposit successful")
        else:
            print("Invalid deposit amount")
    elif transaction == "withdrawal":
        if balance - amount >= 5000 and amount > 0:
            balance = balance - amount
            print("Withdrawal successful")
        else:
            print("Minimum balance requirement not met or invalid amount")
    else:
        print("Invalid transaction type")
else:
    print("Invalid account type")

print("Current balance: ₹", balance)
```

```python
#Get a number n from the user and calculate the sum of all even numbers and the sum of all odd numbers between 1 and n.
n = int(input("Enter a number: "))

even_sum = 0
odd_sum = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i

print("Sum of even numbers:", even_sum)
print("Sum of odd numbers:", odd_sum)
```

```python
'''Get an integer from the user and count the number of digits using a single for loop.
Do not convert the number into a string. Example: 58392 → 5 digits.'''

n = abs(int(input("Enter an integer: ")))
count = 0

for i in range(n):
    count += 1
    n = n // 10
    if n == 0:
        break

if count == 0:
    count = 1

print("Number of digits:", count)
```

```python
#Given a list of distinct numbers, find the second-largest number without using sort() or max().
numbers = [10, 25, 8, 40, 15]

largest = numbers[0]
second_largest = numbers[1]

if second_largest > largest:
    largest, second_largest = second_largest, largest

for num in numbers[2:]:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest:
        second_largest = num

print("Second-largest number:", second_largest)
```

```python
#Get a string from the user and count the vowels and consonants separately. Ignore spaces, digits and special characters.
text = input("Enter a string: ")

vowels = 0
consonants = 0

for ch in text.lower():
    if ch.isalpha():
        if ch in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
'''Get the units consumed by each of five customers using a list. Calculate each customer's bill using these rates:
- First 100 units: ₹2 per unit
- Next 100 units: ₹3 per unit
- Above 200 units: ₹5 per unit'''
units_list = []

for i in range(5):
    units = int(input("Enter units consumed: "))
    units_list.append(units)

for units in units_list:
    if units <= 100:
        bill = units * 2
    elif units <= 200:
        bill = 200 + (units - 100) * 3
    else:
        bill = 200 + 300 + (units - 200) * 5

    print("Units:", units, "Bill: ₹", bill)
```

```python
#Get an integer from the user and reverse its digits using a single for loop. Do not convert it into a string. Example: 12345 → 54321.
n = int(input("Enter an integer: "))

temp = abs(n)
digits = 1

while temp >= 10:
    temp //= 10
    digits += 1

reverse = 0

for i in range(digits):
    digit = abs(n) % 10
    reverse = reverse * 10 + digit
    n = abs(n) // 10

print("Reversed number:", reverse)
```

```python
#Given a list of marks, count the students who passed (50 or above), failed (below 50), and scored distinction (75 or above).
marks = [85, 45, 76, 32, 60, 90, 49]

passed = 0
failed = 0
distinction = 0

for mark in marks:
    if mark >= 50:
        passed += 1
    else:
        failed += 1

    if mark >= 75:
        distinction += 1

print("Passed:", passed)
print("Failed:", failed)
print("Distinction:", distinction)
```
