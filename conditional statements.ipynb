```python
#write a program to print result based on persentage
#85 to 100 is distinction
#60 to 84 is first class
#50 to 59 is second class
#35 to 49 is pass
#<35 is fail
per=float(input("enter the marks"))
if(per<0 or per>100):
    print("invalid")
elif(per>=85):
    print("distinction")
elif(per>=60):
    print("first class")
elif(per>50):
    print("second class")
elif(per>=35):
    print("pass")
else:
    print("fail")
```

```python
#write a program to print its a equilateral triangle,isoceles triangle,scalene triangle,or not a triangle
a=float(input("enter 1st side of triangle"))
b=float(input("enter 2nd side of triangle"))
c=float(input("enter 3rd side of triangle"))
if(a+b<c or a+c<b or b+c<a):
    print("not a triangle")
elif(a==b==c):
    print("equilateral triangle")
elif(a==b or b==c or c==a):
    print("isocelese triangle")
else:
    print("scalar triangle")
```

```python
# program for zero ,+ve,-ve using (nested if)
num=float(input("enter number:"))
if(num!=0):
    if(num>0):
        print(num,"is +ve")
    else:
        print(num,"is -ve")
else:
    print(num,"is zero")
    
```
